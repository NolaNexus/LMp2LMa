## 2. Technology Tradeoff Analysis

# Compare key platform choices


```python
# Technology comparison matrix
tech_choices = {
    "Cloud ML Platform": {
        "pros": ["Managed services", "Auto-scaling", "Preemptible GPUs"],
        "cons": ["Vendor lock-in", "Data transfer costs", "Latency"]
    },
    "On-Prem / Kubernetes": {
        "pros": ["Full control", "Data residency", "Cost at scale"],
        "cons": ["Operational overhead", "Manual scaling", "Higher upfront"]
    },
    "Vertex AI (GCP)": {
        "pros": ["Unified platform", "Auto ML", "Easy scaling"],
        "cons": ["Cost", "Limited customization"]
    },
    "SageMaker (AWS)": {
        "pros": ["Broad tooling", "SageMaker Pipelines", "EC2 integration"],
        "cons": ["Complex pricing", "Steeper learning curve"]
    },
}

for tech, info in tech_choices.items():
    print(f"\n{tech}:")
    print(f"  Pros: {info['pros']}")
    print(f"  Cons: {info['cons']}")
```

## 3. Generate Architecture Decision Records (ADRs)

# ADR template and example

```python
# Create ADRs directory
import os
os.makedirs("architecture/adrs", exist_ok=True)

# ADR 1: Choose Kubernetes over raw EC2 for model serving
adr1 = """
---  
Title: Use Kubernetes (k3d/local) for model serving instead of raw EC2 instances
Status: Accepted
Context: Need to serve multiple models with automatic scaling, GPU sharing, and rollback capability. Team has Kubernetes experience.
Decision: Deploy model serving on Kubernetes with NVIDIA Triton Inference Server and HPA (Horizontal Pod Autoscaling).
Consequences:
  Positive: Auto-scaling, zero-downtime deployments, easy GPU sharing among models.
  Negative: Kubernetes operational complexity, learning curve for team.
Alternatives Considered:
  • Raw EC2 with manual scaling: simpler but no auto-scaling or rollback.
  • Managed service (SageMaker/Vertex AI): vendor lock-in, higher cost at scale.
---  
""".strip()

with open("architecture/adrs/adr001_k8s_serving.md", "w") as f:
    f.write(adr1)
print("ADR 1 written: k8s serving")
```
